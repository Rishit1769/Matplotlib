```python
# =============================================================================
# Bar Charts with Matplotlib - IPL Cricket Data Visualization
# =============================================================================
# This notebook demonstrates vertical and horizontal bar charts, grouped bars,
# and stacked bar visualizations using IPL cricket season data.
# =============================================================================

# --- Core Library Imports ---
import numpy as np                          # Numerical computing (used for bar positioning)
import pandas as pd                         # Data manipulation and CSV handling
import matplotlib.pyplot as plt             # Primary plotting library
import seaborn as sns                       # Statistical data visualization (enhanced styles)
```

```python
# --- Basic Vertical Bar Chart ---
# Demonstrates plt.bar() with categorical x-axis labels.
# Here, 'colors' serves as both the category labels and x-axis ticks.
children = [10, 20, 40, 10, 30]            # Sample values (counts)
colors = ['red', 'blue', 'green', 'yellow', 'black']  # Category labels

plt.bar(colors, children, color='black')    # Plot bars; all bars rendered in black
```

```python
# --- Horizontal Bar Chart ---
# plt.barh() creates horizontal bars instead of vertical ones.
# Useful when category labels are long or when you have many categories.
plt.barh(colors, children, color='black')   # Horizontal bar with same data
```

```python
# --- Bar Chart from CSV Data: Batsman Season Records ---
# Loads IPL season data and plots runs scored by each batsman in 2015.
# The CSV contains batsman names and their runs across 3 seasons.
df = pd.read_csv('batsman_season_record.csv')  # Load batsman season data

plt.bar(df['batsman'], df['2015'])          # Plot 2015 season runs per batsman
```

```python
# --- Grouped Bar Chart: Comparing Multiple Seasons Side by Side ---
# To place bars side by side, we offset each bar group using np.arange().
# Each bar has a width of 0.2 and is shifted left/right to avoid overlap.
# Note: Cell 3 (this cell) plots 2015 twice; Cell 4 fixes the labels.
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width=0.2)  # Left-shifted bar for 2015
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2)          # Center bar for 2016
plt.bar(np.arange(df.shape[0]) + 0.2, df['2015'], width=0.2)  # Right-shifted bar (intended: 2017)
```

```python
# --- Grouped Bars with Custom X-Axis Labels ---
# plt.xticks() replaces numeric indices (0, 1, 2...) with batsman names.
# This makes the chart readable by showing actual player names.
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width=0.2)  # 2015 season bars
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2)          # 2016 season bars
plt.bar(np.arange(df.shape[0]) + 0.2, df['2015'], width=0.2)  # 2017 season bars

# Replace x-tick positions with batsman names
plt.xticks(np.arange(df.shape[0]), df['batsman'])
```

```python
# --- Stacked Bar Chart ---
# The 'bottom' parameter stacks bars on top of each other.
# Each subsequent bar starts where the previous one ends.
# This visualizes the cumulative total across multiple seasons.
plt.bar(df['batsman'], df['2017'])                           # Base layer: 2017 runs
plt.bar(df['batsman'], df['2016'], bottom=df['2017'])        # Middle layer: stacked on 2017
plt.bar(df['batsman'], df['2017'], bottom=(df['2016'] + df['2017']))  # Top layer: stacked on (2016 + 2017)
```
