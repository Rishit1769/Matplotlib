```python
# =============================================================================
# Pie Charts with Matplotlib - IPL Cricket Data Visualization
# =============================================================================
# This notebook demonstrates pie chart creation for proportional data,
# including styling with custom colors, explode effects, shadows,
# percentage labels, and figure export.
# =============================================================================

# --- Core Library Imports ---
import numpy as np                          # Numerical computing
import pandas as pd                         # Data manipulation and CSV handling
import matplotlib.pyplot as plt             # Primary plotting library
import seaborn as sns                       # Statistical data visualization
```

```python
# --- Basic Pie Chart with Labels ---
# plt.pie() creates a pie chart from proportional data.
# The 'labels' parameter adds text labels to each slice.
data = [23, 45, 100, 35, 67, 54]           # Sample values (e.g., marks per subject)
subjects = ['Maths', 'Science', 'English', 'History', 'Geography', 'Computer']  # Category names

plt.pie(data, labels=subjects)              # Create pie chart with subject labels
```

```python
# --- Pie Chart from CSV Data with Percentage Labels ---
# Loads Chris Gayle's historic 175-run innings data and shows the
# run distribution among all batsmen in that match.
# 'autopct' formats percentage values on each slice.
df = pd.read_csv('gayle-175.csv')           # Load Gayle's 175-run match data
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%1.1f%%')  # Pie with 1-decimal percentages
```

```python
# --- Styled Pie Chart with Custom Colors, Explode, and Shadow ---
# Advanced pie chart styling for presentation-quality visualizations.
# - 'colors': Custom color palette for each slice
# - 'explode': Offsets the first slice (Gayle's 175) for emphasis
# - 'shadow': Adds a shadow effect for a 3D appearance
df = pd.read_csv('gayle-175.csv')           # Reload dataset
plt.pie(
    df['batsman_runs'],
    labels=df['batsman'],
    autopct='%1.1f%%',                      # Show percentages with 1 decimal
    colors=['red', 'blue', 'green', 'yellow', 'orange', 'pink'],  # Custom color palette
    explode=[0.1, 0, 0, 0, 0, 0],          # Explode the first slice (Gayle)
    shadow=True                             # Add drop shadow for visual depth
)
```

```python
# --- Exporting Pie Chart to PNG File ---
# plt.savefig() exports the current figure to a file.
# Parameters:
#   - 'pie-chart.png': Output filename (PNG format)
#   - dpi=300: Resolution (300 dots per inch for print-quality output)
# The chart is first recreated, then saved to disk.
df = pd.read_csv('gayle-175.csv')           # Reload dataset
plt.pie(
    df['batsman_runs'],
    labels=df['batsman'],
    autopct='%1.1f%%',
    colors=['red', 'blue', 'green', 'yellow', 'orange', 'pink'],
    explode=[0.1, 0, 0, 0, 0, 0],          # Explode Gayle's slice
    shadow=True                             # Shadow effect
)
plt.savefig('pie-chart.png', dpi=300)       # Save to PNG at 300 DPI
```
