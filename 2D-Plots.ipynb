```python
# =============================================================================
# 2D Line Plots with Matplotlib - IPL Cricket Data Visualization
# =============================================================================
# This notebook demonstrates fundamental line plotting techniques using
# Matplotlib, including customizations for colors, line styles, markers,
# legends, grid lines, and axis limits.
# =============================================================================

# --- Core Library Imports ---
import numpy as np                          # Numerical computing and array operations
import pandas as pd                         # Data manipulation and CSV handling
import matplotlib.pyplot as plt             # Primary plotting library
import seaborn as sns                       # Statistical data visualization (enhanced styles)
```

```python
# --- Basic Line Plot: Hardcoded Data ---
# A minimal example demonstrating plt.plot() with two lists.
# The x-axis represents years, and the y-axis represents price values.
price = [100, 200, 300, 200, 100]           # Sample price data points
year = [2015, 2016, 2017, 2018, 2019]      # Corresponding years

plt.plot(year, price)                       # Plot year vs price as a line chart
```

```python
# --- Line Plot from CSV Data: Virat Kohli's Yearly Runs ---
# Loads IPL data from a CSV file and plots Virat Kohli's annual run totals.
# The 'index' column contains years, and 'V Kohli' contains runs scored.
batsman = pd.read_csv('sharma-kohli.csv')   # Load Sharma-Kohli comparison dataset
plt.plot(batsman['index'], batsman['V Kohli'])  # Plot Kohli's runs over the years
```

```python
# --- Line Plot from CSV Data: Rohit Sharma's Yearly Runs ---
# Similar to the Kohli plot above, but for Rohit Sharma's performance.
batsman = pd.read_csv('sharma-kohli.csv')   # Reload the dataset
plt.plot(batsman['index'], batsman['RG Sharma'])  # Plot Sharma's runs over the years
```

```python
# --- Multi-Series Line Plot: Overlaying Two Batsmen ---
# Combines both batsmen on a single chart for direct visual comparison.
# Matplotlib automatically assigns different colors to each line.
plt.plot(batsman['index'], batsman['V Kohli'])       # Kohli's data series
plt.plot(batsman['index'], batsman['RG Sharma'])     # Sharma's data series
```

```python
# --- Adding Chart Annotations: Title and Axis Labels ---
# Adds descriptive text to the chart for clarity and presentation quality.
plt.plot(batsman['index'], batsman['V Kohli'])       # Kohli's data series
plt.plot(batsman['index'], batsman['RG Sharma'])     # Sharma's data series

plt.title('Runs scored by Kohli and Sharma')  # Chart title
plt.xlabel('Year')                             # X-axis label
plt.ylabel('Runs')                             # Y-axis label
```

```python
# --- Custom Line Colors for Visual Distinction ---
# Assigns explicit colors to each line for better readability.
# Red for Kohli (RCB's signature color), Black for Sharma.
plt.plot(batsman['index'], batsman['V Kohli'], color='red')     # Kohli in red
plt.plot(batsman['index'], batsman['RG Sharma'], color='black') # Sharma in black

plt.title('Runs scored by Kohli and Sharma')
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
# --- Line Style Customization: Dash-Dot Pattern ---
# The linestyle parameter controls the line pattern.
# Options: '-' (solid, default), '--' (dashed), '-.' (dashdot), ':' (dotted)
plt.plot(batsman['index'], batsman['V Kohli'], color='red', linestyle='dashdot')     # Red dashdot for Kohli
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot') # Black dashdot for Sharma

plt.title('Runs scored by Kohli and Sharma')
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
# --- Adding Markers at Data Points ---
# The marker parameter adds visual indicators at each data point.
# Common markers: 'o' (circle), '+' (plus), 'x' (cross), 's' (square), '^' (triangle)
plt.plot(batsman['index'], batsman['V Kohli'], color='red', linestyle='dashdot', marker='+')     # Plus markers for Kohli
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', marker='+') # Plus markers for Sharma

plt.title('Runs scored by Kohli and Sharma')
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
# --- Legend for Multi-Series Charts ---
# The label parameter assigns a name to each line for the legend.
# plt.legend() displays the legend box, typically in the upper-left corner.
plt.plot(batsman['index'], batsman['V Kohli'], color='red', linestyle='dashdot', marker='+', label='Kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', marker='+', label='Sharma')

plt.title('Runs scored by Kohli and Sharma')
plt.xlabel('Year')
plt.ylabel('Runs')

plt.legend()  # Display legend using labels from the 'label' parameters above
```

```python
# --- Controlling Axis Limits ---
# Demonstrates how to clip the visible range using xlim() and ylim().
# This is useful for zooming into specific regions or handling outliers.
# Note: The last price value (1,000,000) is an outlier that gets clipped.
price = [100, 200, 300, 200, 1000000]      # Includes an intentional outlier
year = [2015, 2016, 2017, 2018, 2019]

plt.plot(year, price)

plt.ylim(0, 1000)   # Restrict y-axis to range [0, 1000], hiding the outlier
plt.xlim(2015, 2019) # Restrict x-axis to the data range
```

```python
# --- Adding Grid Lines for Readability ---
# Grid lines help readers estimate values from the chart.
# plt.grid() adds a grid behind the plot lines.
plt.plot(batsman['index'], batsman['V Kohli'], color='red', linestyle='dashdot', marker='+', label='Kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', marker='+', label='Sharma')

plt.title('Runs scored by Kohli and Sharma')
plt.xlabel('Year')
plt.ylabel('Runs')

plt.legend()   # Display legend
plt.grid()     # Enable grid lines for easier value reading
```

```python
# --- Final Complete Chart with plt.show() ---
# plt.show() explicitly renders the figure and displays it.
# In Jupyter notebooks, this is optional (plots auto-display), but
# it is best practice in scripts and production code.
plt.plot(batsman['index'], batsman['V Kohli'], color='red', linestyle='dashdot', marker='+', label='Kohli')
plt.plot(batsman['index'], batsman['RG Sharma'], color='black', linestyle='dashdot', marker='+', label='Sharma')

plt.title('Runs scored by Kohli and Sharma')
plt.xlabel('Year')
plt.ylabel('Runs')

plt.legend()   # Display legend
plt.grid()     # Enable grid lines

plt.show()     # Render and display the complete chart
```
