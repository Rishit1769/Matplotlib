```python
batsman = pd.read_csv('batter.csv')
batsman.head(50)
```

```python
plt.scatter(batsman['avg'],batsman['strike_rate'],color='black',marker = '+')
plt.title('Average vs Strike Rate')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.show()
```

```python
import ssl
from io import BytesIO
from urllib.request import urlopen

tips_url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'
tips = pd.read_csv(BytesIO(urlopen(tips_url, context=ssl._create_unverified_context()).read()))
tips

plt.scatter(tips['tip'], tips['total_bill'],s=tips['size'] )
```
