```python
batsman = pd.read_csv('sharma-kohli.csv') 
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
plt.title('Runs scored by Kohli and Sharma')
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
plt.plot(batsman['index'], batsman['V Kohli'],color='red')
plt.plot(batsman['index'], batsman['RG Sharma'],color='black')

plt.title('Runs scored by Kohli and Sharma')

plt.xlabel('Year')
plt.ylabel('Runs')

```

```python
plt.plot(batsman['index'], batsman['V Kohli'],color='red', linestyle='dashdot')
plt.plot(batsman['index'], batsman['RG Sharma'],color='black', linestyle='dashdot')

plt.title('Runs scored by Kohli and Sharma')

plt.xlabel('Year')
plt.ylabel('Runs')

```

```python
plt.plot(batsman['index'], batsman['V Kohli'],color='red', linestyle='dashdot',marker = '+')
plt.plot(batsman['index'], batsman['RG Sharma'],color='black', linestyle='dashdot',marker = '+')

plt.title('Runs scored by Kohli and Sharma')

plt.xlabel('Year')
plt.ylabel('Runs')

```
